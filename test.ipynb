```python
import numpy as np
```

```python
# An example for the Flickr data. You can change the name for other dataset.
data_name = "flickr"
results_pehe = []
results_msey = []
for i in range(10):
    cur_name = './results/test_results_' + data_name + "_GITE.GITEModel_test_split_" + str(i)  + ".npy"
    cur_re = np.load(cur_name)

    results_pehe.append(cur_re[1])
    results_msey.append(cur_re[0])
```

```python
print(np.mean(results_pehe ** 0.5), np.std(results_pehe ** 0.5))
print(np.mean(results_msey ** 0.5), np.std(results_msey ** 0.5))
```
